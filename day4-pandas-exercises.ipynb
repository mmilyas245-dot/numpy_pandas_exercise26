## Part 3 — Discretization (binning)

**3.1** Create a new column `grade` from `final_marks` using these bands:
`0–50 → F`, `50–60 → D`, `60–70 → C`, `70–80 → B`, `80–100 → A`.
Then show the count of students in each grade.

```python
# TODO 3.1  (hint: pd.cut with bins=[...] and labels=[...])
df["grade"] = pd.cut(df['final_marks'],
                     bins=[0,50,60,70,80,100],
                     labels=['F','D','C','B','A'])
```

```python
df.head()
```

**3.2** Create a column `attendance_band` splitting `attendance` into `low` (0–70), `medium` (70–85), `high` (85–100). Then find the average `final_marks` for each band.
*(What does the result suggest?)*

```python
# TODO 3.2
df["attendance_band"] = pd.cut(
    df["attendance"],
    bins=[0,70,85,100],
    labels=["low","medium","high"]
)
lowavg=df.groupby("attendance_band")['final_marks'].mean().sort_values(ascending=False)
lowavg
# OBSERVATION: what a fantastics coding of python pandas providing such a vast facility of grouping and making new column based on
#  existing one with multiple condition using bins and labels 
```

**3.3** Use `apply()` with a lambda to create `study_level`: `"high"` if `study_hours >= 10`, `"medium"` if `>= 5`, otherwise `"low"`. Then show how many students fall in each level.

```python
# TODO 3.3
df["study_level"] = df["study_hours"].apply(lambda h: "high" if h >=10 else ("medium" if h >=5 else 'low'))
df[["study_hours","final_marks","study_level"]].head(10)
```
