**3.4 — The outlier experiment.** Below are 10 house prices (in lakhs). Compute the mean and median. Then append one mansion priced at 5000 and recompute both. Which measure moved more?

```python
prices = np.array([45, 52, 48, 61, 55, 43, 58, 50, 47, 53])
```

```python
# TODO 3.4
print("BEFORE → mean:",prices.mean(), " median:", np.median(prices))

with_mansion = np.append(prices,5000)              # hint: np.append
print("AFTER  → mean:", with_mansion.mean(), " median:",np.median(with_mansion))

# WHICH MOVED MORE, AND WHY: ...
#mean moved more because the average increase occur in one one the value increase the value of other as well
```
