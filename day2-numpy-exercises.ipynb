## Part 5 — Z-scores, percentiles and quartiles

**5.1** For the exam scores below, compute the z-score of every student.
*(The formula is `(value − mean) ÷ std` — broadcasting doing statistics.)*

```python
exam = np.array([55, 62, 78, 91, 45, 88, 70, 66, 82, 59])
```

```python
# TODO 5.1
z = [(i-(exam.mean())/exam.std()) for i in exam ]
np.round(z, 2)
```

**5.2** A student scored **91**. What is their z-score, and what does that number mean in plain words?

```python
# TODO 5.2
z_91 = exam[3]-(exam.mean()/exam.std())
print(round(z_91, 2))
# MEANING: ...
```

**5.3** Compute the 25th, 50th and 75th percentiles of `exam`. Then compute the median separately and confirm it matches one of them.

**5.4** Compute the **interquartile range** (IQR) of `exam` — that is, Q3 minus Q1. Half of all students fall inside this range.

```python
# TODO 5.4
q1, q3 = (np.percentile(exam,75)),(np.percentile(exam,25))
print("IQR:",q1-q3)
```

**C.3** Given the marks below, replace every score that is **more than 2 standard deviations from the mean** with the median.
*(This is a simple outlier-handling technique you will meet again when cleaning real data.)*

```python
marks = np.array([72, 68, 75, 70, 5, 74, 71, 69, 98, 73])
```

```python
marks.mean()
```

```python
[marks>[(marks.mean()+marks.std())] &marks<[(marks.mean()-marks.std())]]
```

```python
marks.std()
```

```python
# TODO C.3
z = 2*marks.std()
outliers = np.abs(marks-marks.mean())>z         # a boolean mask
print("outliers found:", marks[outliers])

cleaned = marks.copy()
cleaned[outliers] = np.median(marks)
print("cleaned:", cleaned)
```
